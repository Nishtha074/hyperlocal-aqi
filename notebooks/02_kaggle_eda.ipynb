This notebook helps you load a Kaggle 
2015
-2020
 CSV (place `city_day.csv` or `station_day.csv` into `data/raw/kaggle/`) and run quick EDA for Pune.

```python
import os
import pandas as pd

kfile = 'data/raw/kaggle/city_day.csv'
if not os.path.exists(kfile):
    print('Please download the Kaggle dataset and place city_day.csv into data/raw/kaggle/')
else:
    df = pd.read_csv(kfile)
    df['Date'] = pd.to_datetime(df['Date'])
    pune = df[df['City'].str.lower() == 'pune'] if 'City' in df.columns else df[df['city'].str.lower()=='pune']
    cols = [c for c in ['Date','PM2.5','PM10','NO2','SO2','CO','O3','AQI'] if c in pune.columns]
    display(pune[cols].head())
    pune.to_parquet('data/processed/pune_kaggle_2015_2020.parquet')
    print('Saved data/processed/pune_kaggle_2015_2020.parquet')
```
